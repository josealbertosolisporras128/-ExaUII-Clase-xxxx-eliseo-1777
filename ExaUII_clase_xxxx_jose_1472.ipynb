```python
# Software desarrollado por Jose Solis NL 51
# Fecha: 29/abril/26

class sofa:
    def __init__(self, material, plazas, color):
        self.material = material
        self.plazas = plazas
        self.color = color

    def leerdatos(self):
        self.material = input("Ingrese el material del sofa: ")
        self.plazas = input("Ingrese la cantidad de plazas: ")
        self.color = input("Ingrese el color del sofa: ")

    def mostrardatos(self):
        print("Material: ", self.material)
        print("Plazas: ", self.plazas)
        print("Color: ", self.color)

# Creación de objetos

print("--- Sofa numero 1 ---")
sofa1 = sofa("", "", "")
sofa1.leerdatos()
sofa1.mostrardatos()

print("\n--- Sofa numero 2 ---")
sofa2 = sofa("", "", "")
sofa2.leerdatos()
sofa2.mostrardatos()

print("\nSoftware desarrollado por Jose Solis NL 51 - 30/abril/26")
```
